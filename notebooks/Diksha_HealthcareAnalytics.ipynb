## 📐 Section 2: Statistical Analysis

Rigorous statistical testing to validate observations from the EDA.

### Tests Performed
- **ANOVA**: Comparing billing amounts across medical conditions and admission types
- **T-Tests**: Gender-based differences in billing and length of stay
- **Chi-Square**: Association between categorical variables (e.g., condition vs. blood type)
- **Correlation Analysis**: Pearson/Spearman correlations between numeric features

### Interpretation Guide
- **p < 0.05** → Statistically significant result (reject H₀)
- **p ≥ 0.05** → No statistically significant result (fail to reject H₀)

## 📋 Section 4: Summary & Conclusions

### Key Findings
1. **Demographics**: Patient population spans a wide age range with fairly even gender distribution. Blood type O+ is most prevalent.
2. **Medical Conditions**: Certain conditions (e.g., Diabetes, Hypertension) dominate admission records and are strongly associated with emergency admissions.
3. **Financials**: Billing amounts show high variance across insurance providers and hospitals. Significant outliers exist suggesting billing inefficiencies.
4. **Hospital Performance**: A small subset of hospitals and doctors account for disproportionately high patient volumes and revenue.
5. **Length of Stay**: Strongly correlated with billing amount and medical condition severity.

### ML Model Outcomes
- **Billing Predictor** achieved high R² scores — billing can be reliably estimated from clinical and demographic features.
- **Risk Classifier** successfully identifies high-risk emergency admission patients with strong precision-recall balance.
- **Patient Clustering** revealed distinct cohort profiles useful for targeted care pathway design.

### Recommendations
1. **Resource Allocation**: Prioritize staffing and resources around high-admission medical conditions.
2. **Billing Efficiency**: Audit and standardize billing practices across insurance providers.
3. **Targeted Care**: Implement specialized care plans for high-risk clusters identified by K-Means.
4. **Length of Stay Reduction**: Develop discharge optimization protocols for conditions with disproportionately long stays.
5. **Data Enrichment**: Augment the dataset with clinical outcome data for improved predictive accuracy.
